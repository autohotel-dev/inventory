```python
# Upload vehicle_model.onnx first
from google.colab import files
print('Upload vehicle_model.onnx...')
uploaded = files.upload()
print(f'Uploaded: {list(uploaded.keys())}')
```

```python
import onnx2tf
import tensorflow as tf

print('Converting ONNX to TensorFlow...')
onnx2tf.convert(
    input_onnx_file_path='vehicle_model.onnx',
    output_folder_path='tf_model',
    output_signaturedefs=True,
    non_verbose=True
)
print('✓ TensorFlow model saved to tf_model/')
```

```python
import tensorflowjs as tfjs

print('Converting to TensorFlow.js format...')
tfjs.converters.save_keras_model(
    tf.keras.models.load_model('tf_model'),
    'tfjs_model'
)
print('✓ TF.js model saved to tfjs_model/')
```

```python
import os

print('Generated files:')
for root, dirs, files in os.walk('tfjs_model'):
    for f in files:
        path = os.path.join(root, f)
        size = os.path.getsize(path) / 1024
        print(f'  {path} ({size:.1f} KB)')

# Download
from google.colab import files
import shutil

shutil.make_archive('tfjs_model', 'zip', 'tfjs_model')
files.download('tfjs_model.zip')
print('\nDownloaded tfjs_model.zip')
```
